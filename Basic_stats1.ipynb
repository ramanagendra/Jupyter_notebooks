# <h2 style= "color:blue;"><u><b>Standardization of Numerical variables</b></u></h2>

Standardiztion is data processing technique used to transform numerical variables onto a common scale. This is particularly useful when variables have different units or ranges(e.g., volume vs Net Sales  Value). Standardization improves the performance of many statistics and machine learning models by ensuring that no variable dominates due to its scale.

### Concept of Standardization (Z-Score Normalization:

Standardization converts each value into the number of standard deviations it is away from mean.

The formula is: 

z= (x-µ)/(σ)

1. x = original value
2. µ = Mean of the variable
3. σ = standard deviation of the variable
4. z = Standardization value

Interpretation:
1. z=0 --- value equals to mean
2. z>0 --- Value is above the Mean
3. z<0 --- value is below the mean
4. |z|>2 or 3 --- potential outlier


### Why Standardize?

1. Brings all numerical variables to a common scale.
2. Prevents variables with large values from dominating the analysis
3. Improves the performace of algorithm such as K-Means Clustering, PCA(Principal component Analysis), Logistics Regression, SVM ,and Nueral networks etc...

```python
from sklearn.preprocessing import StandardScaler

numerical_cols = [
    'Volume',
    'Avg Price',
    'Total Sales Value',
    'Discount Rate (%)',
    'Discount Amount',
    'Net Sales Value'
]

scaler = StandardScaler()

df_standardized = df.copy()

df_standardized[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

# <h2 style="color:blue;"><u><b>Conversion of Categorical Data into Dummy Variables</b></u></h2>

The Objective of converting categorical data into dummy variables is to transform non numeric categorical features into a numerical format that can be used by Machine Learning algoithms. Most ML algorithms cannot directly process text based categrical values and therefore require numrical inputs

Need for One -Hot Encoding:

The Dataset contains categorical cvariables such as :
1. Day
2. SKU
3. City
4. BU
5. Brand
6. Model
these variables contains text values

So ML algorithms cannot interpret these categories directly. One Hot encoding converts each category ito as eparate binary column containing either : 1-- category present, 0--- category is abscent

| Brand   | Brand_Jeera| Brand_Gear | Brand_Viva |
|---------|---------   |---------   |---------   |
| Jeera   |     1      |   0        |    0       |
| Gear    |     0      |   1        |    0       |
| Viva    |     0      |   0        |    1       |

```python
categorical_cols = [
    'Day',
    'SKU',
    'City',
    'BU',
    'Brand',
    'Model'
]

df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first= True,
    dtype = int
)

```
